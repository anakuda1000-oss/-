```python
%%writefile app.py
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib
import streamlit as st

st.title("散布図の作成")
st.write("Excelファイルを選び、グラフに使う数値列を指定してください。")
excel_file = st.file_uploader("Excelファイルを選択してください", type=["xlsx"])

if excel_file is None:
    st.info("散布図を作るには .xlsx ファイルを選択してください。")
else:
    try:
        data = pd.read_excel(excel_file)
    except (ImportError, OSError, ValueError) as error:
        st.error(f"Excelファイルを読み込めませんでした: {error}")
    else:
        if data.empty:
            st.warning("Excelファイルにデータがありません。")
        else:
            numeric_columns = data.select_dtypes(include="number").columns.tolist()
            if len(numeric_columns) < 2:
                st.error("散布図には数値列が2列以上必要です。")
                st.write("読み込んだ列:", ", ".join(map(str, data.columns)))
            else:
                x_column = st.selectbox("横軸（X）", numeric_columns)
                y_columns = [column for column in numeric_columns if column != x_column]
                y_column = st.selectbox("縦軸（Y）", y_columns)

                ax = data.plot(
                    kind="scatter",
                    x=x_column,
                    y=y_column,
                    title=f"{x_column}と{y_column}の散布図",
                    grid=True,
                )
                figure = ax.figure
                st.pyplot(figure)
                plt.close(figure)
```
